## Les règles de l'île

L'ile est représentée sous formes de cases hexagonales. Il y a différents types de terrains et d'altitudes, que l'on voit sur la carte en fonction de la couleur.

Les types de terrain sont :

- océan : infranchissable
- sable : on peut s'y déplacer assez facilement
- désert : on peut s'y déplacer assez facilement
- herbe : on peut s'y déplacer très facilement
- forêt : on s'y déplace un peu moins vite
- forêt dense : on s'y déplace plus lentement
- marais : le déplacement est très ralenti
- roche : on s'y déplace plus lentement
- neige : le déplacement y est très délicat

L'altitude joue aussi sur les déplacements : on peut grimper de 1 ou 2 unités de hauteur sans difficulté, on peut grimper de 3 ou 4 plus difficilement, on ne peut pas gravir 5 unités d'un seul coup (par exemple, passer d'un case de hauteur 1 à une case de hauteur 6 est impossible, la pente est trop raide) ; en revanche, on peut toujours descendre, quelle que soit la pente.

Concrètement, se déplacer d'une case à l'autre sans changer de hauteur prend 1 unité de temps. Le type de terrain **sur lequel on rentre** et le différentiel d'altitude viennent y ajouter un coût. Pour le type de terrain, les pénalités sont les suivantes :

| Terrain     | Pénalité de déplacement |
| ----------- | ----------------------- |
| Océan       | ∞                       |
| Sable       | 1                       |
| Désert      | 3                       |
| Herbe       | 0                       |
| Forêt       | 2                       |
| Forêt dense | 5                       |
| Marais      | 10                      |
| Roche       | 8                       |
| Neige       | 20                      |

Et pour le différeniel d'altitude :

| Différentiel d'altitude | Pénalité de déplacement |
| ----------------------- | ----------------------- |
| $\leq 0$                | 0                       |
| 1 ou 2                  | 2                       |
| 3                       | 6                       |
| 4                       | 12                      |
| $\geq 5$                | ∞                       |

Pour travailler dans ce carnet, vous avez un fichier `hmap.json` qui contient toutes les informations sur les différentes cases de la carte. On se déplace sur la carte uniquement de case en case voisines. Comme on est sur une carte à hexagones, chaque case a six voisins.

Pour se repérer, on utilise un système de coordonnées axiales (ou obliques), et au lieu de $x$ et $y$, on se repère avec deux axes $q$ et $r$ comme illustré ci-dessous (image de [RedBlobGames](https://www.redblobgames.com/grids/hexagons/)) :
![Schéma montrant l'orientation dans une carte hexagonale](hexagone.png "Hexagones et coordonnées axiales")

Plus simplement, ce qu'il faut retenir est qu'il y a six voisins à la case $(q,r)$, dont les coordonnées sont :
- $(q,r+1)$
- $(q,r-1)$
- $(q+1,r)$
- $(q-1,r)$
- $(q+1,r-1)$
- $(q-1,r+1)$
